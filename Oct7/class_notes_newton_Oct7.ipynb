```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
```

```python
def newton_alg(func, dfunc , x0, tol, maxiter):
    x= x0
    hist = []
    for idx in range(maxiter):
        f, df = func(x) , dfunc(x)
        if abs(df)< 1e-12:
            raise ValueError("df is very small")
        dx = f /df 
        x = x - dx 
        hist.append(x)
        if abs(dx) < tol:
            return x, hist
    raise ValueError("No convergence")

x= sp.symbols('x')
expr = x**2 -2 
df_expr = sp.diff(expr, x)
func = sp.lambdify(x, expr, "numpy")
df_func = sp.lambdify(x, df_expr, "numpy")
x0 =1.0



root, hist_root = newton_alg(func, df_func , x0, tol=1e-6, maxiter=100)
print(f"Root:{root:.6f}")
```
