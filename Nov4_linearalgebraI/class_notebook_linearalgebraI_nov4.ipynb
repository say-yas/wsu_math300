```python
import numpy as np
import sympy as sp
from IPython.display import display
```

## Making matrices

```python
Msp = sp.Matrix([[1, sp.Rational(1,3)],[2, 4]])
display(Msp)

Mnp = np.array([[1, 1/3], [2, 4]], dtype = float)
print(Mnp)
```

```python
Isp = sp.eye(3)
display(Isp)

a , b, c, d = sp.symbols("a b c d")
diagsp = sp.diag(a, b, c, d)
display(diagsp)

zerosp = sp.zeros(2,3)
display(zerosp) 
```

```python
Inp = np.eye(2)
print(Inp)

diagnp = np.diag([ 1, -1 , 2, 3])
print(diagnp)

zeronp = np.zeros((3,2))
print(zeronp)
```

```python
Msp = sp.zeros(3,3)
Msp[0,1] = a
Msp[2,0] = c

display(Msp)
```

```python
Mnp= np.zeros((3,3))
Mnp[0,1] = 2.4
Mnp[2,0] = 1.3

print(Mnp)
```

```python
Asp = sp.Matrix(4,4, lambda i,j: sp.Rational(1, i+j + 1) )
display(Asp)

Asp2= sp.Matrix(3,3, lambda i,j: i+j)
display(Asp2)
```

```python
Anp2 = np.fromfunction(lambda i,j : i+j, (3,3))
print(Anp2)
```

```python
Anp2[2,0]= 10
print(Anp2, "\n")
print(Anp2.T)
```

```python
display(Asp, Asp.T)
display(Asp.is_symmetric())
```

```python
Asp3= sp.Matrix(3,3, lambda i,j: i+j+(i-j)*sp.I)
display(Asp3)
display(Asp3.H)
display(Asp3.is_hermitian, Asp3.is_symmetric())
display(Asp3.det())
```

```python
Anp3 = np.fromfunction(lambda i,j : i+j+(i-j)*1j, (3,3))
print(Anp3)
print(Anp3.conj().T)

print(np.allclose(Anp3, Anp3.conj().T))
print(np.linalg.det(Anp3))
```
