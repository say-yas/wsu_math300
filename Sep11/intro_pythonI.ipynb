# How to print

The `print()` function displays output to the screen. It's the most basic, and most useful, tool for seeing what your code is doing.

A few things demonstrated in the cell below:
- `print("some text")` prints a string.
- You can pass multiple arguments separated by commas, e.g. `print(2, "is a number.")`; `print` will insert a space between them automatically.
- `\n` inside a string creates a new line.
- **f-strings** (`f"x = {x}"`) let you embed the value of a variable directly inside a string: the part in `{}` is evaluated and inserted.
- You can assign multiple variables at once: `x, y = 10, 20`.
- You can put multiple statements on one line by separating them with a semicolon `;`, e.g. `x = 2; z = 5` (though it's usually clearer to use separate lines).


```python
print("Hello, World!")
print(2, "is a number.\n3 is also a number.")

x = 5
print(f"x = {x}")

x,y = 10, 20
print(f"x = {x}, y = {y}")

x=2; z=5
print(f"x = {x}, z = {z}")
```

# Variables

A **variable** is a name that refers to a value stored in memory. In Python, you don't need to declare a variable's type: it's inferred automatically from the value you assign.

The `type()` function tells you what kind of value a variable holds. The basic built-in types shown below are:
- `int`: whole numbers (e.g. `2`)
- `float`: decimal numbers (e.g. `3.14`, or `4.` which is the same as `4.0`)
- `str`: text, written with either double quotes `"..."` or single quotes `'...'`
- `bool`: a truth value, either `True` or `False`


```python
a = 2 # This is an integer
print(a, type(a)) # Print the value of a and its type

b = 3.14 # This is a float
print(b, type(b)) # Print the value of b and its type

c = 4.
print(c, type(c)) # Print the value of c and its type

d = "This is a string"
print(d, type(d)) # Print the value of d and its type

e = 'This is also a string'
print(e, type(e)) # Print the value of e and its type

f = True # This is a boolean
print(f, type(f)) # Print the value of f and its type
```

# Operations

Python supports the standard arithmetic operators:

| Operator | Meaning |
|---|---|
| `+` | addition |
| `-` | subtraction |
| `*` | multiplication |
| `**` | exponentiation (power) |
| `/` | division (always returns a `float`) |
| `//` | floor division (division rounded down to the nearest whole number) |
| `%` | modulo (the remainder after division) |

Notice that combining an `int` and a `float` (e.g. `a + b` where `a` is an `int` and `b` is a `float`) produces a `float` result.


```python
x = a + b # This is an addition operation
print(x, type(x)) # Print the value of x and its type

y= a - b # This is a subtraction operation
print(y, type(y)) # Print the value of y and its type

z = a * b # This is a multiplication operation
print(z, type(z)) # Print the value of z and its type

z2 = a ** b # This is an exponentiation operation
print(z2, type(z2)) # Print the value of z2 and its type

w = c / b # This is a division operation
print(w, type(w)) # Print the value of w and its type

v = c // a # This is a floor division operation
print(v, type(v)) # Print the value of v and its type

u = c % a # This is a modulo operation  
print(u, type(u)) # Print the value of u and its type
```

### Comparison operators

Comparison operators compare two values and return a `bool` (`True` or `False`):

| Operator | Meaning |
|---|---|
| `<` | less than |
| `<=` | less than or equal to |
| `>` | greater than |
| `>=` | greater than or equal to |
| `==` | equal to (note the double `=`; a single `=` is assignment, not comparison!) |

Note that `2` (an `int`) and `2.0` (a `float`) are considered equal in value, even though they are different types.


```python
 a = 2
 b = 2.0
 print(type(a), type(b)) # Print the types of a and b

 x = (a < b) # This is a less than comparison
 print(x, type(x)) # Print the value of x and its type

 x2 = (a <= b) # This is a less than or equal to comparison
 print(x2, type(x2)) # Print the value of x2 and its type

 y = (a > b) # This is a greater than comparison
 print(y, type(y)) # Print the value of y and its type

 y2 = (a >= b) # This is a greater than or equal to comparison
 print(y2, type(y2)) # Print the value of y2 and its type

 z = (a == b) # This is an equality comparison
 print(z, type(z)) # Print the value of z and its type

```

# Containers: Lists

A **list** is an ordered, mutable collection of values, written with square brackets `[...]` and separated by commas. Lists can hold any type of value, and can be changed after they're created (e.g. by adding elements).


## Lists

Some common list operations:
- `len(list1)` returns the number of elements in the list.
- **Indexing** with `list1[i]` retrieves the element at position `i`. Python indexing starts at `0`, so `list1[0]` is the *first* element.
- `list1.append(value)` adds `value` to the end of the list, modifying it in place.


```python
list1 = [1.1, 2.2, 3.2, 4.4, 5.5] # This is a list
print(list1, type(list1)) # Print the value of list1 and its type

l = len(list1) # This is a length operation
print(l, type(l)) # Print the value of l and its type

first_element = list1[0] # This is an indexing operation
print(first_element, type(first_element)) # Print the value of first_element and its type

second_element = list1[1] # This is an indexing operation
print(second_element, type(second_element)) # Print the value of second_element and its type

# append an element to the list
list1.append(6.6) # This is an append operation
print(list1, type(list1)) # Print the value of list1 and its type
```

## Slicing Operation

**Slicing** lets you extract a sub-portion of a list using the syntax `list[start:stop:step]`:
- `start` is the index to begin at (inclusive, default is the beginning of the list).
- `stop` is the index to end at (**exclusive**: the element at this index is *not* included; default is the end of the list).
- `step` is how many elements to move forward each time (default is `1`). A negative step reverses the direction.

Any of these can be omitted: `list1[:]` copies the whole list, `list1[::2]` takes every second element, and `list1[::-1]` reverses the list.


```python
print("The list is:", list1) # Print the value of list1

sublist = list1[1:3] # This is a slicing operation: start index is 1, end index is 3 (exclusive)
print(sublist, type(sublist)) # Print the value of sublist and its type

sublist2 = list1[2:] # This is a slicing operation: start index is 2, end index is the end of the list
print(sublist2, type(sublist2)) # Print the value of sublist2 and its type

sublist3 = list1[:3] # This is a slicing operation: start index is the beginning of the list, end index is 3 (exclusive)
print(sublist3, type(sublist3)) # Print the value of sublist3 and its type

sublist4 = list1[:] # This is a slicing operation: start index is the beginning of the list, end index is the end of the list
print(sublist4, type(sublist4)) # Print the value of sublist4 and its type

sublist5 = list1[::2] # This is a slicing operation: start index is the beginning of the list, end index is the end of the list, step is 2
print(sublist5, type(sublist5)) # Print the value of sublist5 and its type

sublist6 = list1[1::2] # This is a slicing operation: start index is 1, end index is the end of the list, step is 2
print(sublist6, type(sublist6)) # Print the value of sublist6 and its type

sublist7 = list1[::-1] # This is a slicing operation: start index is the end of the list, end index is the beginning of the list, step is -1
print(sublist7, type(sublist7)) # Print the value of sublist7 and its type

sublist8 = list1[3:1:-1] # This is a slicing operation: start index is 3, end index is 1 (exclusive), step is -1
print(sublist8, type(sublist8)) # Print the value of sublist8 and
```

# Useful lists

The `range(start, stop, step)` function generates a sequence of numbers without storing them all in memory at once. It's commonly used for looping a fixed number of times. Like slicing, `stop` is exclusive and `start`/`step` are optional (defaulting to `0` and `1`).

`range()` produces a special `range` object, not a list; wrap it in `list(...)` to see all its values as an actual list.


```python
range_ints = range(10) # This is a range object
print(range_ints, type(range_ints)) # Print the value of range_ints and its type

list_ints_list = list(range_ints) # Convert the range object to a list
print(list_ints_list, type(list_ints_list)) # Print the value of list_ints_list and its type

range2 = range(1, 10, 2) # This is a range object with start, stop, and step
print(range2, type(range2)) # Print the value of range2 and its type

list_range2_list = list(range2) # Convert the range object to a list
print(list_range2_list, type(list_range2_list)) # Print the value of list_range2_list and its type
```

# Iterations

**Loops** let you repeat a block of code multiple times. Python has two main kinds: `for` loops (repeat once per item in a sequence) and `while` loops (repeat as long as a condition is true).


### For loops

A `for` loop iterates over a sequence (like a `range` or a list), running the indented block of code once for each item. The loop variable (`i` below) takes on each value in the sequence in turn.


```python
for i in range(5): # This is a for loop
    x = i + 2
    print(f"i = {i}, x = {x}") # Print the value of i and x
```

You can also loop directly over the elements of a list, rather than over indices. Here `idx` takes each value from the list `[2, 3, 4, 6]` in order.

Note that reassigning the loop variable inside the loop body (`idx = idx - 1`) only changes that variable for the current iteration; it doesn't affect the original list or the next value the loop will use.


```python

for idx in [2,3,4,6]: # This is a for loop over a list
    print(f"idx = {idx} before subtraction") # Print the value of idx
    idx = idx - 1
    print(f"idx = {idx}") # Print the value of idx and y
```

# While

A `while` loop repeats its block of code **as long as** its condition remains `True`. Unlike a `for` loop, the number of iterations isn't fixed in advance; it depends on when the condition becomes `False`.

It's important that something inside the loop eventually makes the condition `False` (here, `x` is decreased by `1` each time), otherwise the loop would run forever!


```python
x = 10
while x > 0:
    x = x - 1
    print(f"x = {x}: ", "The condition is ", x > 0 ) # Print the value of x after subtraction  
```
