# Variables

```python
a = 1 # This is a int variable
print(a, type(a))

b = 3.12 # this is a float variable
print(b, type(b))

c = "Hello" # this is a string variable
d = 'Hello 2'

print(c, type(c))
print(d, type(d))

f = True # this is a boolean variable
print(f, type(f))
```

# Operations

```python
a = 2
b = 3
c = a / b
print(c)

f = a ** b
print(f)

div = b // a
print(div)

remainder = b % a
print(remainder)
```

```python
a = 2
b = 2.
print(type(a), type(b))

print(a == b) # equal operation
print(a != b) # not equal operation
print(a > b) # greater than operation
print(a >= b) # greater than or equal operation
```

# Containers: List

```python
lista = [1.1, 2.2, 3.3, 4.4]
print(len(lista))

print(lista[0]) # first element

l= len(lista)
print(lista[l-1])

lista.append(5.6)
print(lista)
```

List[start(int): stop(int): step(int)]

```python
print(lista)
print(lista[:3])
print(lista[::2]) # every other element

print(lista[::-1]) # reverse the list
```
