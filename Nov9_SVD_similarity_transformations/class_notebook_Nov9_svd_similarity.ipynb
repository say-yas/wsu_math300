```python
import sympy as sp
import numpy as np

from IPython.display import display
```

```python
A = sp.Matrix([[3,1,2],[2,3,-1]])
B = sp.Matrix([[3,1,1],[0,1,0],[1,1,2]])

Anp = np.array(A, dtype=float)
Bnp = np.array(B, dtype=float)
```

# SVD decomposition

A =  U S V^T

```python
U, s, V = A.singular_value_decomposition()
display(U, s, V)

display( U @ s @ V.T == A)
```

```python
singval = A.singular_values()
display(singval)
```

```python
AtA = A.T @ A
display(AtA.eigenvals())
```

```python
U, s, V = np.linalg.svd(Anp, full_matrices=False)
print(U.shape, s.shape, V.shape)

print(np.allclose(Anp, U @ np.diag(s) @ V))
```

```python
U, s, V = np.linalg.svd(Anp, full_matrices=True)
print(U.shape, s.shape, V.shape)

sigma = np.zeros(Anp.shape)
sigma[:len(s), :len(s)] =np.diag(s)
print(sigma)

print(np.allclose(Anp, U @ sigma @ V))
```

```python
s= np.linalg.svd(Anp, compute_uv= False) 
print(s)
```

```python
display(B.eigenvals())

for eigs, multip,basis in B.eigenvects():
    display(eigs, multip)
    for v in basis:
        display(v)
```

```python
w, V = np.linalg.eig(Bnp)

print(np.linalg.matrix_rank(V))

for i in range(len(w)):
    print(w[i].real, V[:, i].real)
```

```python
Cnp = np.array([[3,1,0],[0,3,1],[0,0,3.]])

w, V = np.linalg.eig(Cnp)

print(np.linalg.matrix_rank(V))
for i in range(len(w)):
    print(w[i].real, np.round(V[:, i].real,5))
```
