# Matrices, Symmetry, and Determinants

- **Building matrices**: by hand, from special matrices, element by element, or from a **rule** $a_{ij} = f(i, j)$.
- **Symmetric / Hermitian**: $A^T = A$ / $A^* = A$.
- **Determinant** $\det A$: the factor by which $A$ scales area (volume in 3D). $\det A = 0 \iff A$ is **not invertible**.

```python
import sympy as sp
import numpy as np
from IPython.display import display
```

## Part A. Making matrices

Write a matrix as a **list of rows**.

```python
A = sp.Matrix([[1, 2],
               [3, 4]])                 # SymPy: exact
An = np.array([[1, 2],
               [3, 4]], dtype=float)    # NumPy: floats
display(A)
print(An)
```

**Special matrices:** identity (`eye`), diagonal (`diag`), zeros.

```python
display(sp.eye(3), sp.diag(1, 2, 3), sp.zeros(2, 3))
print(np.eye(3))
print(np.diag([1, 2, 3]))
print(np.zeros((2, 3)))
```

**Fill element by element.** Start from zeros and set entries with `A[i, j]` (row $i$, column $j$; indices start at $0$).

```python
G = sp.zeros(3, 3)
G[0, 1] = 5
G[2, 2] = sp.Rational(1, 3)     # exact 1/3
display(G)

Gn = np.zeros((3, 3))
Gn[0, 1] = 5
Gn[2, 2] = 1/3                  # float 0.333...
print(Gn)
```

**Fill from a rule** $a_{ij} = f(i, j)$. Example: $a_{ij} = i + j$.

- SymPy: `sp.Matrix(rows, cols, lambda i, j: ...)`
- NumPy: `np.fromfunction(lambda i, j: ..., (rows, cols))`

Indices start at $0$. If your rule is written with indices starting at $1$, use `i + 1` and `j + 1` inside the formula.

```python
F = sp.Matrix(3, 3, lambda i, j: i + j)
display(F)

Fn = np.fromfunction(lambda i, j: i + j, (3, 3))
print(Fn)
```

Another rule: $a_{ij} = \dfrac{1}{i + j + 1}$. Use `sp.Rational` in SymPy to keep exact fractions.

```python
R = sp.Matrix(3, 3, lambda i, j: sp.Rational(1, i + j + 1))
display(R)

Rn = np.fromfunction(lambda i, j: 1 / (i + j + 1), (3, 3))
print(Rn)
```

### Symmetric and Hermitian matrices

- **Symmetric**: $A^T = A$ (mirror image across the diagonal).
- **Hermitian**: $A^* = A$, where $A^* = \overline{A}^{\,T}$ is the **conjugate transpose**. For real matrices, Hermitian = symmetric.

**Consequence:** the diagonal entries of a Hermitian matrix must be **real** (since $a_{ii} = \overline{a_{ii}}$).

| | SymPy | NumPy |
|---|---|---|
| transpose | `A.T` | `A.T` |
| conjugate transpose | `A.H` | `A.conj().T` |
| symmetric? | `A.is_symmetric()` | `np.allclose(A, A.T)` |
| Hermitian? | `A.is_hermitian` | `np.allclose(A, A.conj().T)` |

In SymPy, $i$ is `sp.I`; in NumPy it is `1j`. Note `is_hermitian` has **no** parentheses.

```python
Sym = sp.Matrix([[2, 1],
                 [1, 3]])
H = sp.Matrix([[2,        1 - sp.I],
               [1 + sp.I, 3       ]])
display(Sym, H)

print("Sym symmetric?", Sym.is_symmetric(), "  Hermitian?", Sym.is_hermitian)
print("H   symmetric?", H.is_symmetric(),   "  Hermitian?", H.is_hermitian)

Hn = np.array([[2, 1 - 1j],
               [1 + 1j, 3]])
print("NumPy: H Hermitian?", np.allclose(Hn, Hn.conj().T))
```

**Rules can produce symmetric or Hermitian matrices.** The rule $a_{ij} = i + j$ is symmetric because swapping $i$ and $j$ gives the same value. The rule $h_{ij} = (i + j) + (i - j)\,i$ is Hermitian: swapping $i$ and $j$ flips the sign of the imaginary part, i.e. $h_{ji} = \overline{h_{ij}}$.

```python
Hr = sp.Matrix(3, 3, lambda i, j: (i + j) + (i - j)*sp.I)
display(Hr)
print("F  symmetric?", F.is_symmetric())
print("Hr symmetric?", Hr.is_symmetric(), "  Hermitian?", Hr.is_hermitian)
```

## Part B. Determinant

For a $2\times 2$ matrix:
$$ \det\begin{pmatrix} a & b \\ c & d \end{pmatrix} = ad - bc. $$

**Example.** $\det\begin{pmatrix} 3 & 1 \\ 2 & 4 \end{pmatrix} = 3\cdot 4 - 1\cdot 2 = 10$. In SymPy: `A.det()`.

```python
A = sp.Matrix([[3, 1],
               [2, 4]])
A.det()
```

SymPy also works with **symbols**, so it can derive the formula itself:

```python
a, b, c, d = sp.symbols("a b c d")
Ab = sp.Matrix([[a, b],
                [c, d]])
display(Ab)
Ab.det()
```

**3×3 example.** Same command for any size:

```python
B = sp.Matrix([[2, 0, 1],
               [1, 3, 2],
               [1, 1, 2]])
display(B)
B.det()
```

**A singular matrix.** Consider
$$ S = \begin{pmatrix} 1 & 2 & 3 \\ 4 & 5 & 6 \\ 5 & 7 & 9 \end{pmatrix}. $$
Row 3 is the sum of rows 1 and 2, so the rows are dependent and $\det S = 0$.

```python
S = sp.Matrix([[1, 2, 3],
               [4, 5, 6],
               [5, 7, 9]])
S.det()
```

**NumPy pitfall:** the same determinant in floating point is not exactly $0$. Never test `det == 0`; test `abs(det) < 1e-10`.

```python
np.linalg.det(np.array(S, dtype=float))
```

**Determinant of a Hermitian matrix.** Even with complex entries, $\det H$ is always **real** (because $\det H^* = \overline{\det H}$ and $H^* = H$). For $H$ from Part A:

```python
H.det()
```
