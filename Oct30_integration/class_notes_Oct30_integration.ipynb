```python
import numpy as np
import sympy as sp
from scipy import integrate
from IPython.display import display
```

```python
x = sp.symbols("x")
```

```python
intindef = sp.integrate( x* sp.sin(x), x)
display(intindef)
```

```python
intdef = sp.integrate( x* sp.sin(x), (x,0, sp.pi))
display(intdef)
```

```python
sol = sp.integrate(x* sp.sin(x) * sp.cos( x**2) , (x, 0, sp.pi))
# display(sol)
print(sol.evalf())
```

```python
def f(x):
    return x**2 * np.sin(x)

xv = np.linspace(0, np.pi, 21)
yv = f(xv)

solt = integrate.trapezoid(yv, xv)
print("Trapezoid:",solt)

sols = integrate.simpson(yv, xv)
print("Simposon:", sols) 

solfq = integrate.fixed_quad(f, 0, np.pi, n=8)[0]
print("Fixed quad:", solfq)

solq = integrate.quad(f, 0, np.pi)
print("Quad:", solq[0])
errorq= solq[1]
print("Error quad:", errorq)
```
