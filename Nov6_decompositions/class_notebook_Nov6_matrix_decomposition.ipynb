```python
import numpy as np
import sympy as sp
import scipy.linalg as sla 

from IPython.display import display
```

```python
A = sp.Matrix([[1,2,4],[4,5,6],[7,8,1]])
B = sp.Matrix([[0,1,2],[3,4,5],[6,7, sp.Rational(8,9)]])
C = sp.Matrix([[1,3], [2,3],[2,0]])

Anp = np.array(A,  dtype=float)
Bnp = np.array(B, dtype=float)
Cnp = np.array(C, dtype=float)
```

# LU decomposition

$A =  L  U $
$det(A) = det(L)  det(U) = det(U)$

```python
L, U , P = A. LUdecomposition()
display(L,U, P)

display(U.det()== A.det())
display( L @ U == A )
```

```python
display(B)

L, U, P = B.LUdecomposition()
# display(L, U, P)
Pr = sp.eye(3).permute_rows(P)
display(Pr)

display(L @ U == Pr @ B)
display(L @ U == B)
```

```python
P, L, U  = sla.lu(Anp)
# print(P, "\n\n", L, "\n\n", U)
print(np.allclose( A , P @ L @ U))
```

```python
Q, R = A. QRdecomposition()
# display(Q, R)
display(Q.T @ Q)
display(A.det() , R.det())
```

```python
Q, R = np.linalg.qr(Anp)
print(Q, "\n \n", R)
```

```python
print(Cnp, Cnp.shape)
Q, R = np.linalg.qr(Cnp, mode="reduced")
print("reduced mode:", Q.shape, R.shape)

Q, R = np.linalg.qr(Cnp, mode="complete")
print("complete mode:", Q.shape, R.shape)
```
