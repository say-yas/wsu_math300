```python
import numpy as np
import sympy as sp
from scipy.optimize import minimize, minimize_scalar

import matplotlib.pyplot as plt
from IPython.display import display
```

```python
x, y = sp.symbols("x y")
```

f(x) in $[a,b]$
Goal: min/max

$f'(x) =0$ => Solutions are critical points
$f''(x_{crit})$: if >0 local min, if <0 local max
check the values of end points

```python
fsp = x**3 - 2 * x
a = -sp.Rational(3/2)
b = 3

dfsp = fsp.diff(x)
df2sp = fsp.diff(x,2)

sol = sp.solve(dfsp,x)
for s in sol:
    print(f"x={s}")
    print(f"f({s})={fsp.subs(x,s).evalf()}")
    print(f"f''({s})={df2sp.subs(x,s).evalf()}")
    print("-----")

for v in [a,b]:
    print(f"x={v}")
    print(f"f({v})={fsp.subs(x,v).evalf()}")

sp.plot(fsp, (x,a,b), size=(6,3))
```

```python
xs = np.linspace(-1.5, 3 , 101)
fx = xs**3 - 2 * xs

print(f"min at x= {xs[np.argmin(fx)]}, f= {fx.min()}")
print(f"max at x= {xs[np.argmax(fx)]}, f= {fx.max()}")
```

```python
def fn(x):
    return x**3- 2 *x

sol = minimize_scalar(fn, bounds=(-1.5, 3), method="bounded")
print(sol)
print("critical point at : ", sol.x)
print("function value at the critical point:", sol.fun)
```

```python
def nfn(x):
    return -(x**3- 2 *x)

sol = minimize_scalar(nfn, bounds=(-1.5, 3), method= "bounded")
print(sol)
print(f"critical point at :{sol.x}")
print(f"function value ta the critical point:{-sol.fun}")
```
