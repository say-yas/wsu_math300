```python
rng = np.random.default_rng(seed = 43)
N = 1_000_000 
xs = rng.uniform(0, 2, N)
ys = rng.uniform(0, 2, N)

area= 2 * 2 
fs = (xs + ys) * (ys <= xs)

intmc = area * fs.mean()
err = area * fs.std() / np.sqrt(N)
print("Monte Carlo integral:", intmc, "error:", err)
```
