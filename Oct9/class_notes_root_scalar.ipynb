$f(x) = (x^2 + 1) (x-1)$


roots= $\{ \pm i , 1\}$

function_name = lambda paramters: expression

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import scipy.optimize as opt
```

```python
def fpoly(x):
    return (x**2+1)*(x-1)
flam = lambda x: (x**2+1)*(x-1)

xsym=sp.symbols('x')
expr = (xsym**2+1)*(xsym-1)
df_expr = sp.diff(expr, xsym)
df2_expr = sp.diff(expr, xsym, 2)

f = sp.lambdify(xsym, expr, "numpy")
df = sp.lambdify(xsym, df_expr, "numpy")
df2 = sp.lambdify(xsym, df2_expr, "numpy")

print(f"function: {expr}, df: {df_expr}, df2: {df2_expr}")
```

```python
x = np.linspace(-1,2, 100)

fig,ax =plt.subplots(figsize=(3,2))
ax.plot(x, f(x), "b-")
ax.axhline(0, ls="--", color ="silver")

plt.show()
```

```python
root = opt.root_scalar(f, method="ridder", bracket=[0,2])
print(root)
```

```python
root = opt.root_scalar(f, method="newton", x0= 0.5+0.5j, fprime=df)
print(root)
```
